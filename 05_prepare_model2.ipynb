```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# read datasets
df_train = pd.read_csv('../data/processed/df_train.csv')
df_test = pd.read_csv('../data/processed/df_test.csv')
cleaned_df = pd.read_csv('../data/processed/cleaned_data_preprocessing.csv')

print('df_train shape:', df_train.shape)
print('df_test shape:', df_test.shape)
```

### Model 2: Predicting general protective behaviour using all data

```python
# target column
target_col = 'protective_behaviour_binary'

# feature columns
feature_cols = cleaned_df.columns.drop([
    'RecordNo',
    'face_mask_behaviour_scale',
    'protective_behaviour_scale',
    'face_mask_behaviour_binary',
    'protective_behaviour_binary',
    'protective_behaviour_nomask_scale',
    'endtime'
])

# create X
X_train_model_2 = df_train[feature_cols]
X_test_model_2 = df_test[feature_cols]

# create y
le = LabelEncoder()
y_train_model_2 = le.fit_transform(df_train[target_col])
y_test_model_2 = le.transform(df_test[target_col])

# check shapes
print('X_train_model_2:', X_train_model_2.shape)
print('X_test_model_2:', X_test_model_2.shape)
print('y_train_model_2:', y_train_model_2.shape)
print('y_test_model_2:', y_test_model_2.shape)

# save outputs
X_train_model_2.to_csv('../data/model_inputs/X_train_model_2.csv', index=False)
X_test_model_2.to_csv('../data/model_inputs/X_test_model_2.csv', index=False)
pd.DataFrame({'y': y_train_model_2}).to_csv('../data/model_inputs/y_train_model_2.csv', index=False)
pd.DataFrame({'y': y_test_model_2}).to_csv('../data/model_inputs/y_test_model_2.csv', index=False)
```

### Model 2a: Predicting general protective behaviour before mandate

```python
# keep only before-mandate data
train_before = df_train[df_train['within_mandate_period'] == 0].copy()
test_before = df_test[df_test['within_mandate_period'] == 0].copy()

# feature columns for before-mandate model
feature_cols_2a = cleaned_df.columns.drop([
    'RecordNo',
    'face_mask_behaviour_scale',
    'protective_behaviour_scale',
    'face_mask_behaviour_binary',
    'protective_behaviour_binary',
    'protective_behaviour_nomask_scale',
    'endtime',
    'within_mandate_period'
])

# create X
X_train_model_2a = train_before[feature_cols_2a]
X_test_model_2a = test_before[feature_cols_2a]

# create y
le_2a = LabelEncoder()
y_train_model_2a = le_2a.fit_transform(train_before[target_col])
y_test_model_2a = le_2a.transform(test_before[target_col])

# check shapes
print('X_train_model_2a:', X_train_model_2a.shape)
print('X_test_model_2a:', X_test_model_2a.shape)
print('y_train_model_2a:', y_train_model_2a.shape)
print('y_test_model_2a:', y_test_model_2a.shape)

# save outputs
X_train_model_2a.to_csv('../data/model_inputs/X_train_model_2a.csv', index=False)
X_test_model_2a.to_csv('../data/model_inputs/X_test_model_2a.csv', index=False)
pd.DataFrame({'y': y_train_model_2a}).to_csv('../data/model_inputs/y_train_model_2a.csv', index=False)
pd.DataFrame({'y': y_test_model_2a}).to_csv('../data/model_inputs/y_test_model_2a.csv', index=False)
```
