```python
import pandas as pd
from sklearn.model_selection import train_test_split

# read preprocessed dataset
cleaned_df = pd.read_csv("../data/processed/cleaned_data_preprocessing.csv")

# split full data into training and test sets
df_train, df_test = train_test_split(
    cleaned_df,
    test_size=0.2,
    random_state=42,
    stratify=cleaned_df["within_mandate_period"]
)
print("Training set shape:", df_train.shape)
print("Test set shape:", df_test.shape)

# save split datasets
df_train.to_csv("../data/processed/df_train.csv", index=False)
df_test.to_csv("../data/processed/df_test.csv", index=False)

# check before/after mandate sizes in train and test sets
before_train = df_train[df_train['within_mandate_period'] == 0]
after_train = df_train[df_train['within_mandate_period'] == 1]

before_test = df_test[df_test['within_mandate_period'] == 0]
after_test = df_test[df_test['within_mandate_period'] == 1]

print("Before mandate - training set shape:", before_train.shape)
print("Before mandate - test set shape:", before_test.shape)
print("After mandate - training set shape:", after_train.shape)
print("After mandate - test set shape:", after_test.shape)
```
