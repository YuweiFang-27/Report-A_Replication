```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# read datasets
df_train = pd.read_csv("../data/processed/df_train.csv")
df_test = pd.read_csv("../data/processed/df_test.csv")
cleaned_df = pd.read_csv("../data/processed/cleaned_data_preprocessing.csv")
```

### Model 1: Predicting face mask behaviour using all data

```python
# target column
target_col = "face_mask_behaviour_binary"

# feature columns
feature_cols = cleaned_df.columns.drop([
    'RecordNo',
    'face_mask_behaviour_scale',
    'protective_behaviour_scale',
    'face_mask_behaviour_binary',
    'protective_behaviour_binary',
    'endtime'
])

# create X
X_train_model_1 = df_train[feature_cols] 
X_test_model_1 = df_test[feature_cols] 

# create y
le = LabelEncoder()
y_train_model_1 = le.fit_transform(df_train[target_col])
y_test_model_1 = le.transform(df_test[target_col])

# check shape
print("X_train_model_1:", X_train_model_1.shape)
print("X_test_model_1:", X_test_model_1.shape)
print("y_train_model_1:", y_train_model_1.shape)
print("y_test_model_1:", y_test_model_1.shape)

# save outputs
X_train_model_1.to_csv("../data/model_inputs/X_train_model_1.csv", index=False)
X_test_model_1.to_csv("../data/model_inputs/X_test_model_1.csv", index=False)
pd.DataFrame({"y": y_train_model_1}).to_csv("../data/model_inputs/y_train_model_1.csv", index=False)
pd.DataFrame({"y": y_test_model_1}).to_csv("../data/model_inputs/y_test_model_1.csv", index=False)
```

### Model 1a: Predicting face mask behaviour before mandates

```python
# keep only before-mandate data
train_before = df_train[df_train['within_mandate_period'] == 0].copy()
test_before = df_test[df_test['within_mandate_period'] == 0].copy()

# feature columns for before-mandate model
feature_cols_1a = cleaned_df.columns.drop([
    'RecordNo',
    'face_mask_behaviour_scale',
    'protective_behaviour_scale',
    'face_mask_behaviour_binary',
    'protective_behaviour_binary',
    'endtime',
    'within_mandate_period'
])

# create X
X_train_model_1a = train_before[feature_cols_1a]
X_test_model_1a = test_before[feature_cols_1a]

# create y
le_1a = LabelEncoder()
y_train_model_1a = le_1a.fit_transform(train_before[target_col])
y_test_model_1a = le_1a.transform(test_before[target_col])

# check shapes
print('X_train_model_1a:', X_train_model_1a.shape)
print('X_test_model_1a:', X_test_model_1a.shape)
print('y_train_model_1a:', y_train_model_1a.shape)
print('y_test_model_1a:', y_test_model_1a.shape)

# save outputs
X_train_model_1a.to_csv('../data/model_inputs/X_train_model_1a.csv', index=False)
X_test_model_1a.to_csv('../data/model_inputs/X_test_model_1a.csv', index=False)
pd.DataFrame({'y': y_train_model_1a}).to_csv('../data/model_inputs/y_train_model_1a.csv', index=False)
pd.DataFrame({'y': y_test_model_1a}).to_csv('../data/model_inputs/y_test_model_1a.csv', index=False)
```
